Height, Weight -> Numeric


```python
df["Height_cm"] = df["Height"].str.replace("cm", "").astype(int)
df["Weight_kg"] = df["Weight"].str.replace("kg", "").astype(int)

df = df.drop(columns=["Height", "Weight"])
```

Money columns (Eur)->Numeric

```python
import re
import pandas as pd
import numpy as np

def parse_money(value):
    if pd.isna(value):
        return np.nan
    s = str(value).strip().replace("€", "").replace(",", "")
    if s.endswith("M"):
        return float(s[:-1]) * 1_000_000
    if s.endswith("K"):
        return float(s[:-1]) * 1_000
    # sometimes it may already be a plain number-like string
    return float(s)
```

```python
df["Value_EUR"] = df["Value"].apply(parse_money)
df["Wage_EUR"] = df["Wage"].apply(parse_money)
df["Release_Clause_EUR"] = df["Release Clause"].apply(parse_money)

df = df.drop(columns=["Value", "Wage", "Release Clause"])
```

```python
df[["Value_EUR", "Wage_EUR", "Release_Clause_EUR"]].head()
```

Check whats missing


```python
df.isna().sum().sort_values(ascending=False).head(25)
```

```python
#Gotta drop Marking
df = df.drop(columns=["Marking"])
```

```python
categorical_cols = [
    "Loaned From",
    "Joined",
    "Contract Valid Until",
    "Club",
    "Body Type",
    "Real Face",
    "Position",
    "Jersey Number"
]

df[categorical_cols] = df[categorical_cols].fillna("Unknown")
```

```python
df[categorical_cols].isna().sum()
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```
